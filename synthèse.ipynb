## 1. Introduction

### 1.1 Auckland FC et la saison 2024–2025

La saison 2024–2025 de l’A-League Men, championnat professionnel masculin australien auquel participent également des clubs néo-zélandais, marque l’arrivée d’Auckland FC. Malgré ce statut de nouveau venu, Auckland termine à la première place de la saison régulière. Il atteint ensuite les demi-finales de la phase finale, dont le vainqueur reçoit un titre distinct de celui de la saison régulière.

Ce parcours constitue une première raison de s’intéresser à Auckland FC : il est pertinent d’étudier les caractéristiques de jeu d’une équipe qui obtient de tels résultats dès sa saison inaugurale. Les statistiques saisonnières disponibles font également apparaître un profil particulier, en effet Auckland semble être caractérisé par un jeu direct.

Ces observations conduisent à nous intéresser à la façon dont Auckland utilise les espaces proches du but adverse, et plus particulièrement la zone 14. Notre objectif n’est pas d’attribuer les résultats sportifs du club à cette seule zone, mais de décrire les actions qui s’y déroulent et les joueurs qui y contribuent.

L’analyse des événements de match porte sur quatre rencontres d’Auckland FC pour lesquelles nous disposons des données détaillées SkillCorner et des informations nécessaires pour situer les actions sur le terrain.

Le choix de ces matchs offre des situations variées avec des matchs à domicile ou à l'extérieur, trois victoires et une défaite. Auckland a battu les Newcastle Jets (2–0) et Wellington Phoenix (2–1), s’est incliné face à Western United (4–2), puis a remporté le match contre Melbourne Victory (1–0).

Cette diversité permet de comparer les volumes de passes, de tirs et de courses dans la zone 14 selon l’adversaire et le contexte de la rencontre. Elle ne suffit toutefois pas à tirer des conclusions sur l’ensemble des matchs d’Auckland. Les dates, identifiants SkillCorner et résultats détaillés des quatre rencontres figurent dans la section consacrée à la sélection des matchs.

### 1.2 La zone 14 : définition et intérêt tactique

Dans un découpage classique du terrain en 18 zones, la zone 14 correspond à l’espace central situé immédiatement devant la surface de réparation adverse. Elle se trouve entre les deux couloirs, dans l’axe du but. Sa position en fait un espace intéressant pour préparer les dernières actions offensives. Cette zone est particulièrement appréciée des recruteurs et des staffs techniques que ce soit pour établir ce qu'un joueur peut créer offensivement ou que ce soit pour comprendre le fonctionnement de l'animation offensive d'une équipe.

Un joueur peut recevoir le ballon dans cette zone, y préparer une passe vers un partenaire, orienter le jeu ou déclencher une frappe. Les déplacements sans ballon sont également importants : un appel vers la zone 14 peut offrir une solution au porteur, tandis qu’une course depuis cette zone peut accompagner une attaque vers la surface.

**Délimitation géographique de la zone étudiée**

Dans cette étude, nous utilisons le rectangle suivant, cohérent avec les fonctions de traitement des événements :

- **Abscisse** : de `17,5 m` à `longueur du terrain / 2 − 16,5 m` 
- **Ordonnée** : de `−11,33 m` à `+11,33 m`.

L'origine des coordonnées est le centre du terrain et l'axe horizontal positif désigne la direction du but attaqué. Cette définition sert de règle constante pour sélectionner les actions. Elle est un choix opérationnel de découpage spatial, qui doit être explicité car d'autres travaux peuvent retenir des frontières légèrement différentes.


Cependant, l'importance stratégique potentielle de cette zone ne signifie pas que toute action qui y passe devient dangereuse. Pour éviter cette confusion, nous distinguerons la fréquence des actions (combien de passes, tirs et courses ?) et les indicateurs de valeur offensive disponibles (par exemple l'EPV et les événements associés à un tir).

**Comment comptons-nous les actions ?**

- **Passes depuis la zone 14** : possessions de joueur qui se terminent par une passe, avec une position de fin de possession dans le rectangle.
- **Frappes depuis la zone 14** : possessions qui se terminent par un tir, localisées dans ce même rectangle.
- **Passes vers la zone 14** : passes pour lesquelles la possession suivante d'un partenaire est détectée dans la zone, dans un délai de 0 à 30 frames et sans franchir la coupure de mi-temps. Ce sont des **réceptions probables**, pas des transmissions officiellement appariées.
- **Courses vers la zone 14** : événements `off_ball_run` commençant hors de la zone et se terminant dans celle-ci.
- **Courses depuis la zone 14** : événements `off_ball_run` commençant dans la zone et se terminant à l'extérieur.



## 2. Provenance des données et méthodologie

### 2.1 SkillCorner et la collecte d'informations

Les données proviennent du projet SkillCorner Open Data, une mise à disposition de données de football issues d'un système d'analyse fondé sur le suivi des joueurs par vidéo (*broadcast tracking*) et sur des événements calculés à partir de ces observations. Les ressources publiques comprennent notamment des fichiers de matchs, des événements dynamiques (*Dynamic Events*) et des agrégats de saison pour l'A-League Men 2024–2025.

Les événements dynamiques fournissent des informations sur les possessions de joueur, certains déplacements sans ballon et le contexte offensif des actions. Les fichiers de matchs précisent, entre autres, les équipes concernées, les dimensions du terrain et le découpage temporel.

Dans notre environnement de travail Onyxia, les fichiers utilisés ont été lus dans un espace de stockage S3 (`s3://lfaivre/diffusion/`). Les événements de match y sont disponibles sous format **Parquet** et les métadonnées de match sous format **JSON**. Tandis que sur dépôt officiel les données sur les événements de match sont disponibles sous format **CSV** que l'on a du transformer pour faciliter l'utilisation.

La documentation et les données sources sont accessibles sur le dépôt officiel : https://github.com/SkillCorner/opendata.

### 2.2 Deux niveaux de données complémentaires

Les analyses reposent sur deux niveaux d'observation différents.

| Données | Unité d'observation | Variables utilisées | Ce qu'elles permettent d'étudier |
|---|---|---|---|
| **Agrégats de saison — Passing** | Statistiques d'un joueur sur la saison (selon les découpages disponibles) | Passes tentées, réussies, difficiles, progressant à travers une ligne, etc. | Caractéristiques habituelles des passes d'un joueur ou d'une équipe |
| **Agrégats de saison — Off-Ball Runs** | Statistiques d'un joueur sur la saison | Fréquence et types de courses sans ballon | Activité collective et profils de déplacement |
| **Agrégats de saison — Physical** | Statistiques d'un joueur sur la saison | Distances, vitesse, intensité de déplacement | Profil d'activité physique |
| **Événements de match — Dynamic Events** | Une action localisée dans une rencontre | Type d'événement, joueur, équipe, coordonnées, frames, EPV | Où et comment se produisent les actions dans la zone 14 |
| **Métadonnées de match — JSON** | Une rencontre | Équipes, dimensions du terrain, périodes | Situer les événements et représenter le terrain |

La différence essentielle se situe sur le fait que les statistiques de saison résument des tendances sur de nombreuses apparitions, tandis que les événements des quatre rencontres permettent de localiser des actions particulières. Les premières éclairent le profil général d'Auckland tandis que les seconds renseignent sur l'utilisation effective de la zone 14 dans les matchs observés.


### 2.3 Préparation et principes de comparaison

Pour les données saisonnières, les statistiques individuelles sont filtrées pour conserver les joueurs ayant au moins trois matchs observés dans les fichiers utilisés. Les indicateurs déjà rapportés à une durée, par exemple les passes par 30 minutes de possession, sont ensuite agrégés au niveau de l'équipe par des moyennes pondérées. Les poids dépendent du temps d'exposition ou du nombre d'actions selon la variable. Les ratios de passes ou de courses sont distingués des volumes. Les statistiques d'équipe obtenues constituent donc des profils reconstruits à partir des lignes individuelles, et non un tableau officiel directement fourni au niveau collectif.

Ce seuil de trois matchs est un choix pratique d'analyse. Il laisse davantage de joueurs dans l'échantillon, mais peut rendre certains profils moins stables qu'un seuil plus exigeant. Le sens précis des durées et unités doit être vérifié dans la documentation de normalisation SkillCorner.

Pour les données de match, nous utilisons l'identifiant d'Auckland FC (`4177`) pour sélectionner les actions de l'équipe. Les coordonnées sont exprimées en mètres dans le repère des événements, normalisé dans le sens de l'attaque. Nous n'inversons donc pas les coordonnées selon le domicile ou l'extérieur. L'emplacement de la zone dépend de la longueur du terrain fournie par le fichier JSON de chaque rencontre.

## 4. Comment Auckland FC utilise-t-il la zone 14 ?

### 4.1 Vue d'ensemble des quatre matchs

L'étude distingue les actions depuis la zone 14 (passes, frappes et courses qui en partent) et les actions vers cette zone (passes avec réception probable et courses sans ballon qui y entrent). Le tableau rassemble les résultats d'Auckland FC sur les quatre rencontres et les replace dans leur contexte sportif.



Une première différence apparaît entre les matchs remportés contre Newcastle (2–0) et Melbourne Victory (1–0). Face à Newcastle, Auckland enregistre le plus grand volume de passes totales (466), de passes depuis la zone 14 (23), de frappes depuis cette zone (6) et de courses qui en partent (51). Cette rencontre se distingue donc par une utilisation particulièrement fréquente de l'espace central devant la surface adverse. À Melbourne, Auckland gagne également, mais avec seulement 247 passes totales, 4 passes depuis la zone 14 et 14 courses qui en partent. Le succès ne s'accompagne donc pas nécessairement d'une forte activité dans cette zone. La variation moyenne d'EPV des passes depuis la zone 14 y est toutefois la plus élevée des quatre matchs (0,025), contre une valeur proche de zéro face à Newcastle. Il s'agit d'un indicateur de valeur offensive estimée, et non d'une mesure directe des buts créés. Statistiquement, le match contre Melbourne semble avoir été plus difficile pour Auckland expliquant ainsi le fait d'avoir marqué un seul but.

La victoire 2–1 contre Wellington présente un profil intermédiaire, en effet Auckland tente 374 passes, dont seulement 5 depuis la zone 14, mais enregistre 10 passes avec une réception estimée dans cette zone et 21 courses qui en partent. Les volumes d'actions y sont globalement plus faibles que face à Newcastle, alors que le score est plus serré. Ce constat ne permet pas de déterminer si Auckland a davantage souffert pendant le match, mais il souligne que les deux victoires ne correspondent pas à la même utilisation de l'espace.

La défaite 2–4 contre Western United est particulièrement intéressante. Auckland réalise 421 passes au total et 22 courses vers la zone 14, soit le nombre de courses entrantes le plus élevé des quatre rencontres. En revanche, on ne dénombre que 4 passes vers cette zone selon notre estimation, ainsi que 2 frappes depuis celle-ci. Ce décalage peut suggérer des difficultés à faire parvenir le ballon dans l'axe malgré de nombreux déplacements de joueurs. Il ne prouve cependant pas que ces joueurs n'ont pas été servis, en effet les courses et les réceptions estimées ne sont pas appariées individuellement. Surtout, les quatre buts encaissés ne peuvent pas être expliqués par ces seules statistiques offensives.


Les sections suivantes examinent séparément les passes, les frappes et les courses afin de préciser ces contrastes.


### 4.2 Les passes depuis la zone 14

Auckland réalise 23 passes depuis la zone 14 contre Newcastle, contre 5 face à Wellington, 6 contre Western United et 4 face à Melbourne Victory. Contre Newcastle, ces 23 passes représentent environ 4,94 % des 466 passes recensées par notre méthode, une proportion nettement supérieure à celles des autres rencontres (entre 1,34 % et 1,62 %).

Ce volume particulièrement élevé s'observe lors de la victoire 2–0 contre Newcastle. Il contraste avec les quatre passes recensées contre Melbourne Victory, match pourtant remporté 1–0. Auckland a donc pu gagner ces deux rencontres avec des niveaux très différents d'activité de passes depuis la zone 14. De même, les six passes observées contre Western United n'ont pas empêché la défaite 2–4.

Les cartes montrent également une différence de répartition spatiale : contre Newcastle, les points de départ des passes occupent une grande partie du rectangle de la zone 14, alors que contre Melbourne, ils sont beaucoup moins nombreux et regroupés sur quelques positions. Cette observation décrit l'activité d'Auckland dans l'espace, mais elle ne suffit pas à déterminer la dangerosité de chaque passe.

Les indicateurs de variation d'EPV des passes et de lien avec un tir pourront aider à distinguer, lorsque les données sont disponibles, le simple volume de passes de leur contribution offensive estimée.

### 4.3 Les frappes depuis la zone 14

Auckland effectue 6 frappes depuis la zone 14 contre Newcastle, 4 contre Wellington, 2 contre Western United et 4 contre Melbourne Victory. Les cartes de Newcastle et de Melbourne montrent que ces frappes sont déclenchées à plusieurs endroits du rectangle central, sans qu'un seul point de tir soit systématiquement privilégié.

Le contexte des scores permet de nuancer la lecture de ces volumes. Face à Newcastle, Auckland marque deux buts, n'en encaisse aucun et réalise le plus grand nombre de frappes depuis la zone 14. Contre Western United, l'équipe marque également deux buts, mais perd 2–4 : elle n'effectue alors que deux frappes depuis cette zone. Ces chiffres ne permettent toutefois pas de dire que les buts marqués proviennent des frappes recensées dans la zone 14.

Le volume de tirs et la valeur offensive moyenne associée ne suivent pas le même classement. La valeur moyenne de `possession_epv_for_end` est de 0,091 contre Newcastle, 0,105 contre Wellington, 0,076 contre Western United et 0,156 contre Melbourne Victory. L'EPV n'est pas un xG et ne représente pas directement la probabilité de marquer lors du tir. De plus, avec seulement 2 à 6 frappes par match, les moyennes sont sensibles à quelques actions.

### 4.4 Les passes et les courses vers la zone 14

L'accès à la zone 14 peut se faire avec le ballon, par une passe, ou sans ballon, par une course d'un joueur vers cet espace. Les deux indicateurs ne mesurent donc pas la même chose.

Le contraste le plus net apparaît contre Western United, lors de la seule défaite de l'échantillon (2–4). Auckland réalise 22 courses vers la zone 14, soit le total le plus élevé des quatre matchs, mais seulement 4 passes avec réception probable dans la zone. Les cartes montrent de nombreuses courses convergeant vers le rectangle central alors que les trajectoires des passes estimées sont bien moins nombreuses. Malgré deux buts inscrits, cette défaite est peut être due à joueurs non servis dans la zone 14 malgré de nombreux appels.

Face à Newcastle, victoire 2–0, on compte 15 courses et 11 passes estimées vers la zone ; face à Wellington, victoire plus serrée 2–1, 10 et 10 ; face à Melbourne, victoire 1–0, 11 et 5. Même entre les trois victoires, la relation entre ces deux volumes varie. L'activité sans ballon n'est donc pas systématiquement accompagnée d'une progression du ballon vers cet espace.

Les courses et les passes ne sont pas reliées une par une. Une passe peut atteindre la zone sans correspondre à une course détectée, et une course peut être suivie d'une action différente. Il serait donc incorrect d'utiliser ces totaux pour calculer une proportion de courses « servies ». Les arrivées de passes sont par ailleurs estimées à partir de la possession suivante.

### 4.5 Les courses depuis la zone 14

Les courses depuis la zone 14 renseignent sur les déplacements des joueurs qui quittent cet espace après y avoir commencé leur course. Auckland en réalise 51 contre Newcastle, contre 21 face à Wellington, 15 contre Western United et 14 contre Melbourne Victory. Le match gagné 2-0 contre Newcastle se démarque donc très nettement et ces 51 courses peuvent expliquer, avec d'autres indicateurs, la victoire de l'équipe de Auckland.

La carte contre Newcastle révèle des directions variées : plusieurs courses progressent vers l'avant, en direction de la surface adverse, tandis que d'autres se dirigent latéralement ou vers l'extérieur de la zone. La zone 14 semble ainsi servir non seulement de point de départ à des appels offensifs, mais aussi à des déplacements qui proposent différentes solutions de jeu, même si majoritairement, ces appels semblent être dirigés vers la surface de réparation et traduisent donc une envie d'aller au but.

### 4.6 Bilan de l'utilisation de la zone 14

Les quatre matchs mettent en évidence des usages contrastés de la zone 14, sans correspondance simple entre le volume d'actions et le résultat final. Contre Newcastle, Auckland remporte une victoire 2–0 et présente une forte activité dans cet espace : 23 passes, 6 frappes et 51 courses depuis la zone. Face à Wellington, l'équipe gagne plus difficilement 2–1 avec des volumes plus faibles. Contre Melbourne Victory, la victoire est encore plus serrée (1–0) et le nombre de passes depuis la zone est limité à 4, mais l'EPV moyen des possessions associées aux frappes depuis cette zone est le plus élevé de l'échantillon (0,156).

La défaite 2–4 contre Western United apporte un autre contraste : Auckland réalise le plus de courses vers la zone 14 (22), mais seulement 4 passes y aboutissent selon notre estimation. L'équipe marque tout de même deux buts et en encaisse quatre. Ces éléments suggèrent qu'une forte activité de déplacement vers la zone ne se traduit pas nécessairement par davantage de passes reçues dans cet espace, ni par une victoire.

Au total, Auckland inscrit sept buts et en concède cinq sur les quatre rencontres. Ces résultats sportifs donnent un contexte aux statistiques, mais il n'est pas possible d'attribuer les buts marqués ou encaissés aux seules actions analysées en zone 14. Pour approfondir l'étude, il faudra notamment tenir compte des joueurs impliqués, des circonstances de chaque rencontre et du fait que les données de passes vers la zone reposent sur des réceptions estimées.

Ces constats serviront de point de départ à l'identification des joueurs les plus impliqués dans cette zone, puis à une comparaison avec les adversaires.

### 6.5 Bilan de la comparaison et limites

La défaite 4–2 d’Auckland ne correspond pas à une domination de Western United dans tous les volumes d’actions. Auckland identifie davantage de passes et de tirs, mais Western présente un rapport buts/tirs plus élevé. Ce dernier mobilise aussi plus fréquemment la zone 14 dans son jeu de passes : 16 passes depuis cette zone contre 6, et 10 arrivées estimées contre 4, avec une variation moyenne d’EPV plus favorable.

L’analyse par périodes et les cartes confirment que ces différences ne se résument pas à un seul indicateur. Le résultat semble davantage compatible avec une meilleure conversion des tirs et des modalités différentes de progression offensive qu’avec un simple écart de volume de jeu. Il serait toutefois excessif d’en déduire une explication causale certaine : l’échantillon se limite à une rencontre, l’EPV n’est pas un xG et les passes vers la zone 14 restent estimées. Ces observations complètent donc l’étude collective d’Auckland tout en ouvrant la voie à une analyse des joueurs et des circonstances plus précises de cette défaite.
